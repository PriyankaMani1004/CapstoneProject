```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Reading the Dataset
dataset = pd.read_csv(r"D:\AI course Tamil\CapstoneProject\2.Data Preprocessing\Preprocessed_project_risk.csv")
dataset
```

```python
# Separate target first
dependent = dataset["Risk_Level"]
independent = dataset.drop(columns=["Risk_Level"])
```

```python
corr_matrix=independent.corr()
corr_matrix
```

```python
from sklearn.feature_selection import VarianceThreshold
var_selector = VarianceThreshold(threshold=0.01)
X_var = var_selector.fit_transform(independent)

X_var_columns = independent.columns[var_selector.get_support()]
```

```python
X_var_columns
```

```python
X_var
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression(solver='lbfgs')
from sklearn.feature_selection import RFE
X_rfe = RFE(estimator=classifier, n_features_to_select=15)
X_fit = X_rfe.fit(X_var, dependent)
X_rfe_feature=X_fit.transform(X_var)              
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_rfe_feature, dependent, test_size = 1/3, random_state =42)
X_train, X_test, Y_train, Y_test
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression(solver='lbfgs')
classifier.fit(X_train,Y_train)
classifier
```

```python
y_pred=classifier.predict(X_test)
y_pred

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, y_pred)

# print classification report 
from sklearn.metrics import classification_report
clf_report = classification_report(Y_test, y_pred)
```

```python
from sklearn.metrics import f1_score
f1_macro=f1_score(Y_test,y_pred,average='weighted')
print("The f1_macro value for best parameter {}:".format(classifier),f1_macro)
```

```python
print("The confusion Matrix:\n",cm)
```

```python
print("The report:\n",clf_report)
```

```python
from sklearn.metrics import roc_auc_score

y_prob = classifier.predict_proba(X_test)

roc_auc = roc_auc_score(Y_test,y_prob,   multi_class="ovo",   # or "ovo"   
                        average="weighted")

print("ROC AUC:", roc_auc)

```
